```python
import json

# Simulated LLM output parsing (Pre-scripted to mimic LLM reasoning)
def mock_llm_reasoning(step: int, goal: str, observation: str):
    if step == 1:
        return {
            "thought": "I need to check the temperature in Delhi first.",
            "action": "get_weather",
            "args": {"city": "Delhi"}
        }
    elif step == 2:
        return {
            "thought": f"The weather is {observation}. Now I need to calculate total temp if doubled.",
            "action": "calculator",
            "args": {"a": 32, "b": 2, "operation": "multiply"}
        }
    else:
        return {
            "thought": "I have all information required.",
            "action": "FINISH",
            "final_answer": f"Delhi weather is hot, double temp value is {observation}."
        }

# Tool Registry
tools = {
    "calculator": lambda args: calculator(args["a"], args["b"], args["operation"]),
    "get_weather": lambda args: get_weather(args["city"])
}

# Core Agentic Loop
def run_agent(goal: str):
    step = 1
    observation = ""

    print(f"🎯 GOAL: {goal}\n" + "="*40)

    while True:
        # 1. REASON
        decision = mock_llm_reasoning(step, goal, observation)
        print(f"Step {step} | 🧠 Thought: {decision['thought']}")

        if decision["action"] == "FINISH":
            print(f"\n✅ FINAL ANSWER: {decision['final_answer']}")
            break

        # 2. ACT
        action_name = decision["action"]
        tool_fn = tools[action_name]
        print(f"Step {step} | 🛠️ Act: Calling tool '{action_name}' with {decision['args']}")

        # 3. OBSERVE
        observation = tool_fn(decision["args"])
        print(f"Step {step} | 👀 Observe: Tool returned -> '{observation}'\n")

        step += 1

# Execute the agent
run_agent("Find Delhi temperature and calculate 32 multiplied by 2.")
```

```python
# Step 1: Define basic tools as plain functions
def calculator(a: float, b: float, operation: str) -> str:
    if operation == "add":
        return str(a + b)
    elif operation == "multiply":
        return str(a * b)
    elif operation == "divide":
        return str(a / b) if b != 0 else "Error: Division by zero"
    return "Unknown operation"

def get_weather(city: str) -> str:
    # Mocking an external API call
    mock_db = {"Delhi": "32°C, Sunny", "Chennai": "34°C, Humid", "Mumbai": "29°C, Rain"}
    return mock_db.get(city, "City not found")

# Test execution
print("Calculator Test:", calculator(384, 25, "multiply"))
print("Weather Test:", get_weather("Delhi"))
```

```python
# ============================================================
# SIMPLE AI AGENT DEMONSTRATION
# Goal -> Reason -> Act -> Observe -> Reason -> Finish
# ============================================================

# ------------------------------------------------------------
# 1. TOOL: Calculator
# ------------------------------------------------------------

def calculator(a, b, operation):

    if operation == "multiply":
        return a * b

    elif operation == "add":
        return a + b

    elif operation == "subtract":
        return a - b

    elif operation == "divide":
        if b == 0:
            return "Error: Cannot divide by zero"
        return a / b

    else:
        return "Error: Unknown operation"


# ------------------------------------------------------------
# 2. TOOL: Weather
# ------------------------------------------------------------

def get_weather(city):

    # Simulated weather data
    weather_data = {
        "Delhi": "32°C",
        "Mumbai": "29°C",
        "Chennai": "31°C",
        "Bangalore": "24°C",
        "Kolkata": "30°C",
        "roorkee": "32°C"

    }

    return weather_data.get(city, "Weather unavailable")


# ------------------------------------------------------------
# 3. SIMULATED LLM
# ------------------------------------------------------------

def mock_llm_reasoning(step, goal, observation):

    # Step 1:
    # Decide to use the weather tool

    if step == 1:

        return {
            "thought": "I need to check the temperature in Delhi first.",
            "action": "get_weather",
            "args": {
                "city": "Delhi"
            }
        }

    # Step 2:
    # After observing the temperature,
    # decide to use the calculator

    elif step == 2:

        return {
            "thought": "I received the weather information. Now I will calculate 32 multiplied by 2.",
            "action": "calculator",
            "args": {
                "a": 32,
                "b": 2,
                "operation": "multiply"
            }
        }

    # Step 3:
    # Finish the task

    else:

        return {
            "thought": "I have all the information required.",
            "action": "FINISH",
            "final_answer": "Delhi temperature is 32°C and 32 multiplied by 2 is 64."
        }


# ------------------------------------------------------------
# 4. TOOL REGISTRY
# ------------------------------------------------------------

tools = {

    "calculator": lambda args:
        calculator(
            args["a"],
            args["b"],
            args["operation"]
        ),

    "get_weather": lambda args:
        get_weather(
            args["city"]
        )
}


# ------------------------------------------------------------
# 5. AGENT LOOP
# ------------------------------------------------------------

def run_agent(goal):

    step = 1
    observation = ""

    print("=" * 60)
    print("AI AGENT STARTED")
    print("=" * 60)

    print("GOAL:", goal)
    print()

    while True:

        # ----------------------------------------------------
        # REASON
        # ----------------------------------------------------

        decision = mock_llm_reasoning(
            step,
            goal,
            observation
        )

        print("STEP", step)
        print("Thought:", decision["thought"])


        # ----------------------------------------------------
        # FINISH
        # ----------------------------------------------------

        if decision["action"] == "FINISH":

            print()
            print("FINAL ANSWER:")
            print(decision["final_answer"])

            break


        # ----------------------------------------------------
        # ACT
        # ----------------------------------------------------

        action_name = decision["action"]

        print("Action:", action_name)
        print("Arguments:", decision["args"])


        # Find the required tool

        if action_name not in tools:

            print("ERROR: Tool not found:", action_name)
            break

        tool_function = tools[action_name]


        # ----------------------------------------------------
        # OBSERVE
        # ----------------------------------------------------

        observation = tool_function(
            decision["args"]
        )

        print("Observation:", observation)

        print("-" * 60)

        step += 1


# ------------------------------------------------------------
# 6. RUN THE AGENT
# ------------------------------------------------------------

run_agent(
    "Find the Delhi temperature and calculate 32 multiplied by 2."
)
```
